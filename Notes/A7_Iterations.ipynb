```python
''' For-loops '''

solvents_dict = {
    "water": 100,
    "ethanol": 78,
    "N,N-dimethylformamide": 153,
    "dichloromethane": 40,
    }

for k, v in solvents_dict.items():
    print(f"The boiling point of {k} is {v} degree Celcius.")
```

```python
for i in [1, 2, 3]:
    print(i)

for _ in [4, 5, 6]:     # Do something unrelated to i
    print('Hello!')

# Do not change the iterable
```

```python
for k, v in solvents_dict.items():
    if k == "water":
        continue
    if v == 153:
        break
    print(f"    Key: {k}, Value: {v}")
```

```python
for i in range(10):
    if i == 3:
        print("Aborting the loop")
        break
else:
    print("`i` was never 3")    # Executed when the loop is not interrupted by a break
```

```python
for i in range(10):
    if i == 3:
        print("Aborting the loop")
        break
    else:
        print("`i` was never 3")    # else statement is within the loop
```

```python
''' range and enumerate '''

# range(start, stop, step) gives an iterable of integers
# range can be indexed

for i in range(10):
    # start = 0 (inclusive)
    # end = 10 (exclusive)
    # step = 1
    print(i)
```

```python
for i in range(10, 111, 10):
    print(i, end = ' - ')
```

```python
# enumerate(iterable, start=0) extracts the element and according index at the same time
# enumerate cannot be indexed: it is an iterator

for c, i in enumerate(["water", "ethanol", "dichloromethane"]):
    print(f'index {c}: {i}')
```

```python
''' While-loop '''

# the loop is runed under certain condition

x = list(range(-10, 11))        # Create x and E as lists
E = [element ** 2 - 2 for element in x]
print(x)
print(E)
```

```python
i = 0

while E[i] > 0:         # stops when E crosses the zero line
    print(f"E({x[i]}) = {E[i]}")
    i += 1

    if i == len(x):     # stops when all elements are runed
        break

if i != len(x):
    print(f"E({x[i]}) = {E[i]}")
    print(f"E(x) crosses the zero line between x = {x[i-1]} and x = {x[i]}.")
else:
    print("E(x) does not cross the zero line.")
```

```python
l = [2, 1, 3, 7, 5, 6]
i = 0

while True:               # loop until break because "True" is always True
    print("Let's loop")

    if l[i]**2 >= 8:
        break

    i +=1
```

```python
i = 0

while i < len(l)-1:                     # when l[i] is not the last element
    if l[i] > l[i+1]:                   # when l[i] is larger than the next element
        l[i], l[i+1] = l[i+1], l[i]     # a, b = b, a switches position of a b
    i += 1

print(l)
```

```python
''' Iterators and Generators '''

# an iterator is an object that return its elements only once
# e.g. enumerate
# functions that create iterators is generator
# e.g. zip()

g = (i for i in 'abcde')        # expression for a generator
while True:                     # 'an expression of x' for x in 'a iterable'
    try:                        
        print(next(g))          # return the next item from the iterator
    except StopIteration:
        break
```
