```python
import logging

fname = 'Working_Sample/loggingexample.log'
logging.basicConfig(
    filename=fname,                     # Name of file
    filemode="w",                       # "w" = writing mode, "a" = appending mode
    format='%(asctime)s %(message)s',   # Put time before actural logging message
    datefmt='%m/%d/%Y %I:%M:%S %p',     # Format of time : month/day/year hour:minute:second am(or pm)
    level=logging.DEBUG,                # which logging level is printed DEBUGE < INFO < WARNING < ERROR < CRITICAL
)
```
