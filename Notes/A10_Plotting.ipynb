```python
import matplotlib.pyplot as plt     # A simple way to plot
import matplotlib as mpl            # A more complex but flexible bundle
```

```python
plt.plot(range(10))     # plot() takes at least a sequence of y-values.
```

```python
# Generate the lists of x and y values

X = [x * 0.1 for x in range(-10, 11)]
Y = [x ** 2 for x in X]

plt.plot(
    X, Y, 
    color = "grey", linestyle = "--", 
    marker = "o", markerfacecolor = "b", markeredgecolor = "k"
)
```

```python
plt.axhline(1, alpha = 0.5)
# Horizontal line at y=1; transparency set to 0.5

plt.plot(X, Y,
         color = "grey", linestyle = "-",
         marker = "o", markerfacecolor = "g", markeredgecolor = "k",
         label = "LABEL"
)

# Plot headings and labels
plt.title("Green Circles on Parabola")
plt.xlabel("$x$")       # $...$ Displays the string as a math symbol
plt.ylabel("$y$")

# Which y-ticks to give a number
plt.yticks((0, 1))

# Set y-limit
plt.ylim(None, 1.3)

# Generate legend, defaults to read out the labels
plt.legend()
```

```python
fig = plt.gcf()     # Get the current figure, creates one if nothing is found
print(type(fig))
ax = plt.gca()
```

```python
fig, ax = plt.subplots()
# We use pyplot only to create Figure and Axes in the first place

# plot y vs x unsing pink pentagons (marker = "p", which means pentagons)
pink_pentagons = ax.plot(
    X, Y,
    linestyle=" ",
    marker="p", markerfacecolor="#e377c2", markeredgecolor="k",
    label="p"
    )

# plot -y vs x unsing cyan plus (marker = "P" which means plus)
cyan_crosses = ax.plot(
    X, [-y for y in Y],
    linestyle=" ",
    marker="P", markerfacecolor="#17becf", markeredgecolor="k",
    label="P"
    )

# Plot nothing, but create a line object with crosses (marker = "X")
cyan_xs = ax.plot(
    [], [],
    linestyle=" ",
    marker="X", markerfacecolor="#17becf", markeredgecolor="k",
    label="X"
    )

# Add a custom legend
# Use manually specified lines (handles) and labels
ax.legend(
    handles=[cyan_xs[0], pink_pentagons[0], ],  # switched
    labels=["Another P", "Another p"],  # changed
    frameon=False,
)

# Go through all xticks labels and make them bold face
for ticklabel in ax.get_xticklabels():
    ticklabel.set_fontweight("bold")
```

```python
fig, Ax = plt.subplots(2, 1)    # Generate axis in 2 rows, 1 column

upper = Ax[0].plot(             # Ax[0] refers to the first plot
    X, [Y * (-1) ** i for i, Y in enumerate(Y, 1)],
    linestyle = " ", marker = "o", markerfacecolor = '#ab3867', markeredgecolor = "k",
    label = "upper"
)

lower = Ax[1].plot(
    X, [-Y * (-1) ** i for i, Y in enumerate(Y, 1)],
    linestyle = " ", marker = "d", markerfacecolor = "#80d46b", markeredgecolor = "k",
    label = "lower"
)

for ax in Ax:                   # for every axis in Ax that we generated
    ax.set_xticks(())           # Empty label on x-axis
    ax.set_yticks((0,))         # Only label 0 on y-axis
    ax.legend()

fig.subplots_adjust(hspace=0.2) # Adjust the space between subplots
```
