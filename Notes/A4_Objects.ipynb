```python
number = 14.27      # Define a variable
print(number)       
print(id(number))    # Print the memory address of the variable
del number          # Delete the variable
print(number)      
```

```python
"""
Object = attributes (variables) + methods (functions)
Dot-notation can access the attributes.
""";

number = 12.5
print(number.real)  # Access the real part of the number
print(number.imag) 
print(number.is_integer())  # Check if the number is an integer
print(number.as_integer_ratio() ) # Get the integer ratio of the number
```

```python
x = 14
dir(x)  # List all attributes and methods of the object x
```

```python
y = 'Hello Berlin!'
help(y.lower)  # Get help on the lower() method of the string object y
"""
No () after lower because we are not calling the method, just getting help on it.
""";
```
