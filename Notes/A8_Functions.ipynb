```python
def repeat_print(s, i=2):
    """Repeated calls of print()
    
    Arguments:
    s: strings object to print, positional argument
    i: times of repeat, optional keyword argument
    
    Returns: None
    """

    for _ in range(i):
        print(s)

# Call of repeat_print
repeat_print("Hello", 4)
```

```python
def colomb_interact(r, q1=1.602e-19, q2=1.602e-19, k=9e9):
    """Calculate Colomb interaction
    
    Uses F = (k * q1 * q2) / r**2
    
    Positional arguments:
    r (float): radius
    
    Keyword arguments:
    q1, q2 (float): charges of particles.
        Defaults to elemental charges.
    k (float): force constant (N m^2 C^-2).
        Defaults to vacuum.
    
    Returns: result of calculation
    
    """

    return (k * q1 * q2) / r**2

R = [r * 0.1 * 1e-9 for r in range(1, 11)]              # must have a number 1 before e

for r_ in R:
    print(f"F({r_:.2e}) = {colomb_interact(r_):.2e}")   # .2e means 2 decimal place
```

```python
def use_local():
    """Local varialbles are prior to the global ones.
    The variables are first defined and then called.
    """

    x = 2
    print(x)

x = 1
use_local()     # Local variable is used
print(x)        # Global variable is used
```

```python
def can_do_this():
    """Local varialbles or result generated are discarded,
    but they can be used by return or printed by print().
    """

    return string.replace("ABC", "XYZ")

string = "ABC"
string = can_do_this()
print(string)
```

```python
def can_also_do_this():
    """A global variable is modified, this causes no error.
    The codes here are cleaner as list are mutable but strings are not.
    """

    list_.append(1)

list_ = []
can_also_do_this()
print(list_)
```

```python
"""or we can do this in a better way
"""

def now_can_do_this():
    global string   # the global variable is used instead of creating a local one.
    string = string.replace("ABC", "XYZ")

string = "ABC"
now_can_do_this()
print(string)
```

```python
def demonstrate_scopes():

    x = "I am local to 'demonstrate_scopes'"    # does not change global x

    def modify_global():
        global x                                # changes global x
        x = "I was changed by 'modify_global'"

        y = "I am local to 'modify_global'"
        print(y)

    print(x)
    modify_global()                             # executed only when called

x = "I am global"
print(x)
demonstrate_scopes()                            # called here
print(x)
```

```python
def demonstrate_scopes():
    x = "I am local to 'demonstrate_scopes'"

    def modifiy_non_local():
        nonlocal x          # modify x from demonstrate_scopes
        x = "I was changed by 'modify_non_local'"

    def modify_global():
        global x            # modify global x
        x = "I was changed by 'modify_global'"

    print(x)
    modifiy_non_local()
    print(x)
    modify_global()
    print(x)

x = "I am global"
print(x)
demonstrate_scopes()
print(x)
```
