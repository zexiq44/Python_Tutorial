```python
data = []
# Empty list to store the data

with open("Working_Sample/output.txt", "r") as file6:
    for i, line_ in enumerate(file6):
        # i (integer): line index starting from 0
        # line_ (string): content of the line

        if line_.startswith(("#", "@")):
            continue
            # If the line start with # or @, 
            # Skip the line and do not execute following try-loop

        try:
            # Try to add the second column of the line into data
            data.append(float(line_.split()[1]))
            # Split the line in whitespaces as elements in a list
            # Append the element with index 1 in the list
            
        except (IndexError, ValueError):
            # IndexError: no second colomn or indexing fails
            # ValueError: the second column cannot be converted to a float

            print(f"Can't read from line {i}")
            # Print where the errors occur

print(data)
```
