```python
""" Use python to claculate the Coulomb interaction.""";
r=1e-9          # m = 1 nm, Distance
ke=9e9          #N * m^2 / C^2, Coulomb's constant
q1=q2=1.602e-19 #C, Charge of an electron
F=ke*q1*q2/r**2 #N, Coulomb's Law
print("Coulomb interaction:",F,"N")
""" The exponential is xey, where x is the number and y is the exponent. """;
```

```python
x=float('nan')
print(x, type(x))
""" NaN is Not a Number, which is a special floating point value
 that is not equal to any number, including itself. """;

print(x+1)
""" NaN is kept in calculations, can act as a spaceholder for undefined values.""";
```

```python
y=float('inf') # Positive infinity
print(type(y))

z=float('-inf') # Negative infinity
print(type(z))
```

```python
print("Infinity + 1 =", y+1)
print("-Infinity + 1 =", z+1)
print("Infinity + (-Infinity) =", y+z)
```

```python
""" The type of number can be set by 
int()
float()
complex()""";
print(type(int(1.5)))
print(type(float(1)))
print(type(complex(1)))

print(type(int(3+4j)))
""" A type error occurs when the type of the object is
not permitted for the operation."""
```
