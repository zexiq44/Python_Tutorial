```python
list1 = [1, 'Hello', 'Thank you', 2.0]
print(list1[0])
print(list1[-1])
print(list1[1:-1])
print(list1[::2])
print(list1[::-1])
print(list1.index(1))
```

```python
list1[2] += '!!'        # Add '!!' to the string at index 2
print(list1)
```

```python
list1.append('New Item')  # Add a new item to the end of the list
print(list1)
```

```python
list2 = [1, 2, 3, 4, 5]
list1.extend(list2)     # Add all items from list2 to the end of list1
print(list1)
```

```python
list1.insert (0, 'First Item')  # Insert a new item at index 0
print(list1)
```

```python
deleted = list1.pop(2)  # Remove the item at index 2 and return it
print(deleted)
print(list1)
```

```python
list3 = [ 's', 'o', 'r', 't', 'e', 'd' ]
list3.sort()  # Sort the list in place and return None
print(list3)
```

```python
print(list1 + list2)  # Concatenate two lists
print(list3 * 3)  # Repeat the list three times
```

```python
""" 2. Dictionaries """;

dic1 = {'Name': 'Ash',
        'Age': 20,
        'City': 'Manchester'
        }
print(dic1['Age'])
```

```python
dic1 ['Year of Study'] = 2023   # Add a new key-value pair to the dictionary
print(dic1)
```

```python
dic1.update({'Age': 23, 'City': 'Berlin'})  # Update existing key-value pairs
print(dic1)
```

```python
delete = dic1.pop('Year of Study')  # Remove a key-value pair from the dictionary
print(dic1)
print(delete)
```

```python
print(dic1.keys())
print(dic1.values())
print(dic1.items())
```

```python
""" 3. Tuples """;

solvents = ('Water',
            'Ethanol', 
            'Acetone', 
            'Methanol')
print(solvents)
```

```python
# Tuples are sequences, so the elements can be acessed by index and slicing.
print(solvents[0]) 
print(solvents[1:3])  # Access elements from index 1 to 2
print(solvents.index('Acetone'))
```

```python
# Tuples can be different types of objects, but they are immutable.
bp = ("water", 100,
      "ethanol", 78,
      "N,N-dimethylformamide", 153,
      "dichloromethane", 40,)

bp[0] = 12
```

```python
# Tuples can be added or multiplied, but the result is a new tuple.
print( (1, 2, 3) + ('4', '5', '6') )
print( (1, 2, 3) * 2 )
```

```python
# If the tuple has only one element, a comma must be added after the element.
print(('1', ) * 10)
print(('1')*10)
```

```python
""" 4. Sets """;

# Sets are unordered collections of unique elements.
set1 = {4, 2, 3, 5, 3}
print(set1)
```

```python
# Sets are mutable.

set1.add(6)     # Add an element.
print(set1)

set1.update([7, 8, 9])  # Add elements from a list.
print(set1)

set1.remove(2)  # Remove an element.
set1.discard(10)  # Remove an element if it exists, do nothing if it doesn't.
print(set1)
```

```python
# + and * are not defined for sets.
print({1, 2, 3} | {'1', '2', '3'})  # Combine two sets
print({1, 2, 3} & {2, 3, 4})  # Common elements of two sets
print({1, 2, 3} - {2, 3, 4})  # Elements in the first set but not in the second
print({1, 2, 3} ^ {2, 3, 4})  # Elements in either set but not both
```

```python
""" 5. Type Conversion """;

string_list = list('Hello')  # Convert a string to a list of characters
print(string_list)

string_tuple = tuple('Hello')  # Convert a string to a tuple of characters
print(string_tuple)

string_set = set(string_tuple)  # Convert a tuple to a set of unique characters
print(string_set)

string_dict = dict(zip(string_set, string_list))  # Convert a set and a list to a dictionary
print(string_dict)
```
