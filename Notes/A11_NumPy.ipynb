```python
import numpy as np

list1 = [1, 3, 5]
list2 = [0.0, 2.0, 4.0]

A = np.asarray(list1)
B = np.asarray(list2)
print(A, type(A), A.dtype)
print(B, type(B), B.dtype)
# While lists can have different types of elements, the objects stored in a array
# need to have same type.
```

```python
A = np.array([1, 3, 5])     # We can iterate over the elements in an array
for i in A:
    print(i**2)


print(A)                    # Or access them by indexing
print(A[0])
print(A[-1])
print(A[:2])
print(A[1:-1])
print(A[::2])
print(A[::-1])
```

```python
A = np.array([1, 3, 5, 7])      # Generate an array
B = [0, 3]                      # Create a list
print(A[B])                     # Print elements in A that has indeces given in B
```

```python
A = np.array([1, 3, 5, 7])
print(A[A > 1])                 # Test the array against a condition to get a boolean
                                # sequence for indexing
```

```python
A = np.array([1, 3, 5, 7])
print(A + 5)                    # The operation is done element-wise
print(A * 2.0)
```

```python
A = np.array([1, 3, 5, 7])
B = np.array([0, 0.2, 4.0, 6.5])
print(A + B)        # The operations are also done element-wise between arrays
print(A - B)        # The arrays need to be in same length
print(A * B)
print(A / B)
```

```python
res = 5     # set a resolution

r = np.linspace(0, 1, res)
theta = np.linspace( 0, np.pi, res)
phi = np.linspace(-np.pi, np.pi, res)

R, T, P = np.meshgrid(r, theta, phi)    # np.meshgrid build possible combinations of r, theta and phi
                                        # R stores the value of r for all combination (5*5*5)


def f(r, theta, phi):
    # Transform spherical to cartesian coordinates
        
    return (r * np.sin(theta) * np.cos(phi),
            r * np.sin(theta) * np.sin(phi),
            r * np.cos(theta))

np.shape(f(R, T, P))    # When groups of number are input, np.shape() is used to tell the shape of the numbers
                        # 3D shape , each group has 5 elements (5*5*5)
```

```python
a = np.array([1, 2, 3])
b = np.array([0, 4, 0])
M = np.array([[3, 2, 1],
              [1, 2, 3]])

# Inner product of vectors == dot product
print(np.inner(a, b))   # For 1D arrays .inner() calculates the scalar (dot) product = 1*0 + 2*4 + 3*0 = 8
print(np.dot(a, b))     # np.dot() for 1D vectors, same as the dot product here
print(a @ b)            # Matrix multiplication, same as the dot product here
print(np.tensordot(a, b, axes=1))  # tensor dot product, same as dot product for 1D arrays with axes=1
```

```python
# Multiplication of a vector and a matrix
print(np.dot(M, a))     # Multiplication of matrix and vector
print(np.dot(a, M.T))   # M.T transpose the row and column of M to  
                        # [[3, 1], 
                        # [2, 2], 
                        # [1, 3]]

# Cross product of vectors
print(np.cross(a, b))   # Cross product of a, b, returns a vector that is perpendicular to a and b
```

```python
# Outer product of vectors
print(np.outer(a, b))               # Elements in a mutiply every elements in b to produce a matrix
print("\n")
print(np.tensordot(a, b, axes=0))
```

```python
A = np.asarray([1, 3, 5, 7])
B = np.asarray([0, 2.0, 4.0, 8.5])

print(np.subtract(A, B))        # Elements in A - according elements in B
print("\n")
print(np.subtract.outer(A, B))  # Elements in A - every elements in B
```

```python
C = np.array([
    [1, 2, 3],
    [4, 5, 6],
    ])

A = np.array([1, 2, 3])

print("shape C = ", np.shape(C))
print("shape A = ", np.shape(A))
print("\n")

print("C - A = \n", C - A)      # To be substract, A need to have same number of rows
print("\n")

print("shape C.T = ", np.shape(C.T))
print("add an axis before A\n", A[np.newaxis], np.shape(A[np.newaxis]))     # Add an axis before, A become (1, 3)
print("\n")                                                                 # C.T - A cannot be done

print("add an axis after A\n", A[:, np.newaxis], np.shape(A[: np.newaxis])) # Add an axis after, A become (3, 1)
print("\n")
print("C.T - A[:, np.newaxis] = \n", C.T - A[:, np.newaxis])                # C.T - A can be done
```
