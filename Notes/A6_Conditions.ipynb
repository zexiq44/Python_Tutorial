```python
print(type(None))    # No value
print(type(True))    # Boolean value
```

```python
# Empty values are converted to False, while non-empty values are converted to True
print(bool(1))    # Not empty
print(bool(0))    # Empty
print(bool("Hello"))    # Not emptys
print(bool(""))    # Empty
print(bool(None))    # Empty
print(bool([]))    # Empty
```

```python
# In calculations,True is treated as 1 and False is treated as 0
print(True + True)    # 2
print(True + False)    # 1
print(True * False)    # 0
```

```python
# Operators

a = 10
b = 20

print(a == b )   # True if a is equal to b
print(a != b)    # True if a is not equal to b
print(a < b)     # True if a is less than b
print(a > b)     # True if a is greater than b
print(a <= b)    # True if a is less than or equal to b
print(a >= b)    # True if a is greater than or equal to b
print('\n')

c = True
d = False
print(c & d)     # True if c and d are both true
print(c | d)     # True if c or d is true
print(c ^ d)     # True if only one of cd is true

```

```python
# == / != operator evaluates VALUE EQUALITY
# Is / is not operator evaluates OBJECT EQUALITY

print(True == 1)
print(False == 0)
print(True is 1)
print(False is 0)
print('\n')

print(a is b)          # True if a is identicle to b
print(a is not b)      # True if a is not identicle to b
print(c and d)         # True if c and d are both true
print(c or d)          # True if c or d is true
print('\n')
```

```python
# Although bool(None) = False, None it self does not equal to False
print(False == False)
print(False is False)
print(bool(None))
print(None == False)
print(None is False)
```

```python
# and / or operator

print(1 and 2)      # If True return the last True value
print(2 and 1)
print('\n')

print(1 and 0)      # If False return the first False value
print( 0 and 1)
print(None and 0)
print('\n')

print(1 or 2)       # If True return the first True value
print(2 or 1)
print(None or 0)    # If False return the last value
print(0 or None)

""" Notes
and: Find the first falsy, if no, return the last value.
or: Find the first truthy, if no, return the last value.

Priority: not > and > or
The result of and will become a value of 'or' statement.
The position of first result depends on the position of 'and' statement.
"""
```

```python
# in operator
print(1 in [1, 2, 3])       # True if an object is a part of another one
```

```python
# if, elif, else

A = ['Aspirin', 'Ethanol']

if "Aspirin" in A:
    print('I can use this.')

elif 'Ethanol' in A:
    print('ok I can also use this.')

else: print('No I cannot do this.')
```

```python
# try, except, finally

B = (1, 2, 3)

try:
    B.append(4)

except AttributeError:
    pass

print(B)
```

```python
try:                                    # try: where error may occur
    A.append('DMF')

except AttributeError:                  # except: when error occor
    print('This cannot be done.')

else: print('DMF is appended to A.')      # else: when NO error occur

finally: print(A)                       # no matter if error occur
```

```python
# or modify using isinstance: to see if the object is the type

if isinstance(B, list):
    B.append(4)
    print('4 is appended to B.')

else:
    print('This cannot be done.')

print(B)
```
