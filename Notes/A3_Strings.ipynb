```python
s = "ABCDEFG"
print(s)        # print the whole string
print(s[0])     # print the first character of the string
print(s[-1])    # print the last character of the string
print(s[:3])    # print the first three characters of the string
print(s[2:5])   # print the third to fifth characters of the string
print(s[2:])    # print the third character to the end of the string
print(s[::2])   # print every second character of the string
print(s[::-1])  # print the string in reverse order
print(s[10])    # this will raise an IndexError because the index is out of range
"""s[x:y:z] is used to slice a string, where x is the starting index,
y is the ending index, and z is the step size. If x is omitted,
 it defaults to 0. If y is omitted, it defaults to the length of the string. If z is omitted, it defaults to 1. Negative indices can be used to count from the end of the string."""
```

```python
print("=" * 14)     # Strings can be multiplied
print("a" + "b")    # Strings can be added
print("a""b")       # Strings can be concatenated without the + operator
```

```python
print("ABCDEFG\n".replace("A", "a"))  # replace A with a

print("ABCDEFGA".strip("A"))  # remove A from the beginning and end of the string
print("ABCDEFGA".lstrip("A"))  # remove A from the beginning of the string
print("ABCDEFGA".rstrip("A"))  # remove A from the end of the string

print("ABCDEFG".count("A"))  # count the number of A's in the string

print("\nABCDEFG".lower())  # convert the string to lowercase
print("abcdefg".upper())  # convert the string to uppercase
```

```python
result = 5.267
print( f"The result is {result:.2f}")
"""The f-string can add variables to a string in a curly brace {{}}. 
 .2f means 2 decimal places."""

n = 1
print(f"\nmydata_{n:0>5}.txt") 
# :0>5 means the number will be padded with zeros to a width of 5 characters.

n = 2
print(f"\nmydata_{n:0>5}.txt")

n =99999
print(f"\nmydata_{n:0>5}.txt")
```
