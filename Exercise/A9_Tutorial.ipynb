```python
"""
1 Write a set of functions that can be used to maintain a shopping-list saved to a 
file on disk. We want to represent the shopping list in memory as a dictionary with 
items to buy as keys and quantities as values. On disk we need a text file with two 
columns (items, quantity). In the first line, add a comment explaining the content 
(starting with a #). We need the following functions (names are of course free to 
change):
"""

"""
write: the function should get a dictionary and write the current state of the 
shopping-list to disk.
"""

def write_in(dict_, file_):
    # Write the dictionary of shopping list into a file.
    # dict_ (dict): the dictionary

    with open(file_, "w+") as file_1:
        # Create a text file and open it in writing mode.

        file_1.write("# This file is a shopping list with the name and the number of the items.\n")

        for key, value in dict_.items():
            file_1.write(f"{key} {value}\n")
            # Write the shopping list into the file.

        file_1.seek(0)
        # find the line with index 0

        print("Following content is written:\n\n", 
            file_1.read())
            # Print the result of this function.


dict_shopping = {
    "Apple" : 3,
    "Orange" : 4,
    "Mango" : 6,
    "Onion" : 2
}
# The dictionary that stores the shopping list.
# keys (string): name
# values (integer): number

write_in(dict_shopping, "Working_Sample/shopping_1.txt")
```

```python
"""
read: the function should read a file from disk in the expected format and return a 
shopping-list dictionary. The function needs to handle the situation in which no file 
is found to read from.
"""


def read_dict(file_):
    # Read the content in a file and put the shopping list in a dictionary.
    # file_ (string): Route to open the file, ect. "Working_Sample/shopping_1.txt"
    
    dict_read = {}
    # Define an empty dictionary to store the text

    try:
        with open(file_, "r") as file_2:
            # Open the selected file in read mode.

            for line_ in file_2:
                if line_.startswith("#"):
                    continue
                    # Skip the lines of title or comments

                dict_read [line_.split()[0]] = int(line_.split()[1])
                # Add the item into dict_read.

    except IndexError:
        print("Missing value")

    except ValueError:
        print("Invalid number")

        print("Following content is read:\n\n",
          dict_read)
            # Print the result of the function.

    except FileNotFoundError:
        print("The file does not exist.")
        # When the file does not exist.

    return dict_read

read_dict("Working_Sample/shopping_1.txt")
```

```python
"""
add_item: the function should get a new item we need to buy (string) and a quantity 
(integer; let’s neglect units here for the sake of simplicity). The item should be 
added to our dictionary and the file on disk should be updated automatically. The 
function needs to know the current state of the shopping-list (the location of the 
file on disk). Advanced If an item we want to add to the list is already in there, 
add the quantities.
"""

def add_item(name_, quantity_, file_=None):
    # Add a item to the dictionary and the file.

    # Variables:
    # name_ (string): name of the item.
    # quantity_ (integer): quantity of the item.
    # file_ (file): the file wish to read.

    # Here we use the file "Working_Sample/shopping_1.txt"

    if not file_:
         file_ = "Working_Sample/shopping_1.txt"
    
    dict_read = read_dict(file_)

    if name_ not in dict_read:
            dict_read [name_] = quantity_
            # Add a new item to the dictionary.

    else:
        dict_read.update({name_ : (quantity_ + dict_read[name_])})
        # Update the quantity of the item.

    write_in(dict_read, file_)
    return dict_read

print(add_item("Carrot", 1))
```

```python
"""
Advanced cross_off_item: the function should get an item we have bought and a 
quantity. Subtract the quantity from the corresponding item in the list. Remove the 
item completely, if we have bought enough. Print a message, if the item was not in 
the list in the first place.
"""


def cross_off_item(item_, quantity_, file_=None):
    # Delete the item that is already bought.

    # item_ (string): name of the item.
    # quantity_ (integer): quantity of the item.
    
    # Here we use the dictionary dict_shopping and 
    # the file "Working_Sample/shopping_1.txt"

    if not file_:
         file_ = "Working_Sample/shopping_1.txt"

    dict_read = read_dict(file_)

    if item_ in dict_read:
        # The item is in the dict: decrease the amount or delete it.

        if quantity_ >= dict_read[item_]:
            # Delete the item as bought enough.
            dict_read.pop(item_)
            print(f"{item_} is removed from the shopping list.")

        else:
            # Decrease the ammount of the item that is bought.
            dict_read.update({item_ : (dict_read[item_] - quantity_)})
            print(f"{dict_read[item_]} of {item_} is still required.")        

    else:
        # The item is not in the dict.
        print("The item is not in the shopping list.")

    write_in(dict_read, file_)


cross_off_item("Orange", 3)
```

```python
"""
Advanced pretty_print: the function should display the content of the shopping-list 
dictionary in a nicer format. Print for each item the quantities after a newline and 
indented by four spaces.
"""

for keys_, values_ in dict_shopping.items():
    print(f"{keys_}:\n\t{values_}\n".expandtabs(4))
```
