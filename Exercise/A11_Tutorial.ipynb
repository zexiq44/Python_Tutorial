```python
"""
1 Write a function that takes three vectors 𝐚, 𝐛, 𝐜 in 2D (NumPy arrays) and checks 
whether they form a triangle, i.e. if 𝑎+𝑏=±𝑐.
"""


a = [1, 2]
b = [3, 4]
c = [4, 6]

def triangle(A, B, C):
    # Check whether vectors A, B, and C can form a triangle.
    # A, B and C are vectors in list or array.

    import numpy as np

    a = np.asarray(A)
    b = np.asarray(B)
    c = np.asarray(C)
    # Generate an array for each factor
    
    if (np.all(a + b - c == 0) or 
        np.all(a + b + c == 0) or
        np.all(a - b + c == 0) or
        np.all(-a + b + c == 0)):
        print(f"{A}, {B} and {C} can form a triangle.")
    # If the sum of two sides equals the third side, print "can form"
    
    else:
        print(f"{A}, {B} and {C} cannot form a triangle.")
    # If none of the sum of two sides equals the third side, print "cannot form"


triangle(a, b, c)


"""
Solution
"""

import numpy as np
def is_triangle(a, b, c):
    """Check if a, b, c form a triangle

    Args:
        a, b, c (numpy.ndarray): Input vectors

    Returns:
        True or False
    """

    if np.allclose(a + b, np.absolute(c)):  # np.allclose() check whether the elements are proximately equal
        return True                         # np.absolute() turns the elements to absolute value
    return False                            # True / False return is better to store the output
is_triangle(
    np.array([0, 1]),  # a
    np.array([1, 0]),  # b
    np.array([1, 1])   # c
    )
```

```python
"""
2 Consider the vector 𝐯 = (0 0 1)^T and the rotation matrix
𝐑 = (1 0 0)
    (0 0 -1)
    (0 1 0)

Rotate 𝐯 by 𝐑 to obtain 𝐯′
Compute the scalar product of 𝐯 and 𝐯′
By which angle did we rotate along which axis?
"""

import numpy as np

v = np.array([0, 0, 1])
R = np.array([[1, 0, 0],
             [0, 0, -1],
             [0, 1, 0]])

print("v is ", v)
v_ = np.dot(R, v)
print("v after rotation is ", v_)

print("v.T * v_ = ", np.dot(v, v_))
print("R is a 90 degree rotation around x-axis.")
```

```python
"""
3 Consider the second Laplace spherical harmonic
Compute 𝑌 on a grid of 10 points per dimension (0≤𝜃≤𝜋, 0≤𝜙<2𝜋).
"""

res = 10

theta = np.linspace(0, np.pi, res)
phi = np.linspace(0, np.pi*2, res)

T, P = np.meshgrid(theta, phi)
print(T, "\n")
print(P, "\n")
print(np.shape((T, P)))



"""
Solution
"""

theta = np.linspace(0, np.pi, 10)
phi = np.linspace(0, 2*np.pi, 10)
print(r"$\theta$", theta)
print(r"$\phi$", phi)

def Y(theta, phi):
    return 1/2 * np.sqrt(3 / (2*np.pi)) * np.sin(theta) * np.exp(-1j * phi)

T, P = np.meshgrid(theta, phi)

y = Y(T, P)
print(y)
print(y.shape)
```
