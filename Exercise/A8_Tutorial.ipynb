```python
"""
2 Write a function that computes the cumulative sum of values in a sequence and
    returns the result as a list. Demonstrate that it works. Hint: You can use a list
    to pass a sequence to a function.
"""

def add_cumulate(list_):
    """Calculate the cumulative sum of each element in a iterable.
    
    Returns: a list of cumulative sum
    """

    list_cumulate = []
    i = 0
    while i < len(list_):
        list_cumulate.append(sum(list_[:i+1]))
        i += 1
    return list_cumulate

list1 = [1, 2, 3, 4, 5]
print(add_cumulate(list1))


"""
Solution
"""
def cumsum(seq):
    """Compute cumulative sum

    Args:
       seq: Sequence of numbers

    Returns:
       Cumulative sum as list
    """

    result = []
    for i, _ in enumerate(seq, 1):
        result.append(
            sum(seq[:i])
            )
    return result

print(cumsum(range(10)))
```

```python
"""
Advanced 1 
Write a function exp(x) that takes a float x as an argument and returns e𝑥 approximated 
as the Taylor series up to order 𝑛 around the expansion point 𝑎=0. Make the order 𝑛 an 
optional argument that defaults to 𝑛=5. You can call factorial(n) from within exp(x) to 
calculate the factorials."""

def exp(x, n=5):
    """Calculate approximated e^x using Talor series.
    
    Variables:
    x: the e^x to calculate.
    n: order of the series, defaults to 5.
    """

    total = 0

    def calculate_factorial(a):
        """Calculate the factorial value of an integer x.
        """

        n = a
        f = 1
        while n > 0:
            f *= n
            n -= 1
        return f

    for i in range(0, n+1):
        total += (x**i) / calculate_factorial(i)

    return total

print(exp(8))
```
